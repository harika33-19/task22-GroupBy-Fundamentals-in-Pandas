```python
import pandas as pd

data = {
    "Name": ["Ravi", "Priya", "Arun", "Sneha", "Kiran", "Divya"],
    "Department": ["CSE", "ECE", "CSE", "ECE", "CSE", "ECE"],
    "Gender": ["Male", "Female", "Male", "Female", "Male", "Female"],
    "Marks": [80, 90, 70, 85, 95, 75],
    "Attendance": [85, 90, 80, 95, 88, 92]
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df.groupby("Department").groups)
```

```python
for department, group in df.groupby("Department"):
    print("\nDepartment:", department)
    print(group)
```

```python
print(df.groupby("Department")["Marks"].sum())
```

```python
result = df.groupby("Department")["Marks"].agg(
    ["sum", "mean", "count", "min", "max"]
)

print(result)
```

```python
result = df.groupby(
    ["Department", "Gender"]
)["Marks"].mean()

print(result)
```

```python

import pandas as pd

# Create sample data
data = {
    "Name": ["Ravi", "Priya", "Arun", "Sneha", "Kiran", "Divya"],
    "Department": ["CSE", "ECE", "CSE", "ECE", "CSE", "ECE"],
    "Gender": ["Male", "Female", "Male", "Female", "Male", "Female"],
    "Marks": [80, 90, 70, 85, 95, 75],
    "Attendance": [85, 90, 80, 95, 88, 92]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

print("\n1. Total Marks:")
print(df.groupby("Department")["Marks"].sum())

print("\n2. Average Marks:")
print(df.groupby("Department")["Marks"].mean())

print("\n3. Student Count:")
print(df.groupby("Department")["Name"].count())

print("\n4. Minimum Marks:")
print(df.groupby("Department")["Marks"].min())

print("\n5. Maximum Marks:")
print(df.groupby("Department")["Marks"].max())

print("\n6. All Statistics:")
print(df.groupby("Department")["Marks"].agg(
    ["sum", "mean", "count", "min", "max"]
))

print("\n7. Group by Department and Gender:")
print(df.groupby(["Department", "Gender"])["Marks"].mean())

print("\n8. Average Marks and Attendance:")
print(df.groupby("Department")[["Marks", "Attendance"]].mean())
```
